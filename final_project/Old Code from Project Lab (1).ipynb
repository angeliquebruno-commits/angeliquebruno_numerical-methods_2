```python
# Old Code from Fall 25' Project Lab

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parameters
omega = 1.0  # natural frequency

# System of equations
def pendulum(y, t, omega):
    theta, theta_dot = y
    dydt = [theta_dot, -omega**2 * np.sin(theta)]
    return dydt

# Initial conditions
y0 = [2.0, 0.0]  # theta(0) = 2 rad, theta'(0) = 0

# Time points
t = np.linspace(0, 20, 2000)  # 2000 steps between 0 and 20

# Solve ODE
sol = odeint(pendulum, y0, t, args=(omega,))

# Extract theta and theta_dot
theta = sol[:, 0]
theta_dot = sol[:, 1]

# Plot theta and theta_dot
plt.figure(figsize=(12, 6))
plt.plot(t, theta, label=r'$\theta(t)$', color='b')
plt.plot(t, theta_dot, label=r'$\dot{\theta}(t)$', color='r')
plt.xlabel('Time [s]')
plt.ylabel('Angle [rad] / Angular velocity [rad/s]')
plt.title('Nonlinear Pendulum: Angle and Angular Velocity')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import imageio.v3 as iio
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq
import skimage
from skimage.color import rgb2hsv
from skimage.measure import regionprops, label

# --- 1. CONFIGURATION ---

# Define the full paths for all your videos in a list:
VIDEO_FILES = [
    # Change all backslashes (\) to forward slashes (/)
    r'C:/Users/Angelique/Downloads/IMG_3824.mov', 
    # If you have other videos, make sure they also use forward slashes!
]

# ... (function definitions) ...

# --- MAIN ANALYSIS LOOP START ---
# Loop through the list of video file paths
for video_path in VIDEO_FILES: 
    print(f"\n--- Analyzing Video: {video_path} ---")
    
    # Resetting variables for the new video
    time_data = []
    blue_positions = []
    red_positions = []
    
    # Read the video and extract metadata
    # The fix is here: use 'video_path' instead of 'VIDEO_FILE'
    reader = iio.imopen(video_path, 'r') 
    meta = reader.metadata()
    # ... (rest of the code continues inside this loop) ...]

# Define normalized HSV color ranges (Hue: 0-1, Saturation/Value: 0-1)
# **CRITICAL: Adjust these values if tracking fails!**
# Blue (Hue is typically around 0.6 - 0.7)
LOWER_BLUE = np.array([0.60, 0.50, 0.20]) 
UPPER_BLUE = np.array([0.70, 1.00, 1.00])

# Red (Hue wraps around 0/1)
LOWER_RED_1 = np.array([0.00, 0.50, 0.20]) 
UPPER_RED_1 = np.array([0.05, 1.00, 1.00]) 
LOWER_RED_2 = np.array([0.95, 0.50, 0.20]) 
UPPER_RED_2 = np.array([1.00, 1.00, 1.00])


# --- 2. DIODE TRACKING FUNCTION ---
def track_diode(frame, lower_hsv, upper_hsv):
    """Isolates a color in the frame and returns the centroid (x, y)."""
    hsv_frame = rgb2hsv(frame)
    
    # Create the base color mask
    mask = np.all(hsv_frame >= lower_hsv, axis=2) & np.all(hsv_frame <= upper_hsv, axis=2)
    
    # Handle Red's dual range 
    if lower_hsv[0] == 0.00 and upper_hsv[0] == 0.05:
        mask2 = np.all(hsv_frame >= LOWER_RED_2, axis=2) & np.all(hsv_frame <= UPPER_RED_2, axis=2)
        mask = mask | mask2

    # Label connected regions and find the largest one (the diode)
    labeled_img = label(mask)
    regions = regionprops(labeled_img)
    
    if regions:
        largest_region = max(regions, key=lambda r: r.area)
        cy, cx = largest_region.centroid # Centroid returns (y, x)
        return cx, cy
    
    return None, None

# --- 3. VIDEO PROCESSING AND DATA COLLECTION ---

# Read the video and extract metadata
reader = iio.imopen(VIDEO_FILES, 'r')
meta = reader.metadata()
fps = meta.get('fps', 30.0) # Use 30 FPS as a safe default

time_data = []
blue_positions = []
red_positions = []

print(f"Starting analysis at {fps:.2f} FPS...")

for frame_count, frame in enumerate(reader.iter_data()):
    current_time = frame_count / fps
    
    # Track both diodes
    bx, by = track_diode(frame, LOWER_BLUE, UPPER_BLUE)
    rx, ry = track_diode(frame, LOWER_RED_1, UPPER_RED_1) 

    # Store data if both diodes are found
    if bx is not None and rx is not None:
        blue_positions.append(bx)
        red_positions.append(rx)
        time_data.append(current_time)
        
    if frame_count % 100 == 0:
        print(f"Processed frame {frame_count}...")

reader.close()
print(f"Analysis complete. Total frames processed: {frame_count + 1}")

# Convert lists to NumPy arrays
time_data = np.array(time_data)
blue_positions = np.array(blue_positions)
red_positions = np.array(red_positions)

# Calculate Displacement (relative to the average equilibrium position)
blue_displacement = blue_positions - np.mean(blue_positions)
red_displacement = red_positions - np.mean(red_positions)

# --- 4. DATA VISUALIZATION AND ANALYSIS ---

## 📌 Plot 1: Positions as a Function of Time 

# This shows the positions of the blue vs the red diode as a function of time.
plt.figure(figsize=(10, 6))
plt.plot(time_data, blue_displacement, label='Blue Diode (Upper Leg) Displacement', color='blue')
plt.plot(time_data, red_displacement, label='Red Diode (Bottom Leg) Displacement', color='red')
plt.title('Horizontal Displacement of Pendulum Legs Over Time (Beating Motion)')
plt.xlabel('Time (s)')
plt.ylabel('Horizontal Displacement (pixels)')
plt.grid(True)
plt.legend()
plt.show() 


## 🔬 Frequency Measurement and Non-Sinusoidal Evidence

# --- Frequency Analysis (FFT) ---
N = len(time_data)
T = 1.0 / fps # Sample spacing

# Apply FFT to the Red Diode data 
yf_red = fft(red_displacement)
xf = fftfreq(N, T)[:N//2] # Frequency bins (only positive)

# Calculate the power spectrum
power_spectrum_red = 2.0/N * np.abs(yf_red[0:N//2])

# Find the two largest peaks (The Normal Modes)
top_two_indices = np.argsort(power_spectrum_red)[-2:][::-1]
f_mode_1 = xf[top_two_indices[0]] # Lower frequency mode (Symmetric/In-phase)
f_mode_2 = xf[top_two_indices[1]] # Higher frequency mode (Anti-Symmetric/Out-of-phase)


## 📊 Plot 2: Power Spectrum (FFT)

# This fulfills "get a measure of the frequencies" and "show that they are clearly not sineosodial".
plt.figure(figsize=(10, 6))
plt.plot(xf, power_spectrum_red, label='Red Diode Power Spectrum', color='red')
plt.title('Frequency Spectrum (FFT) of Pendulum Motion: Non-Sinusoidal Evidence')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Power/Magnitude')
plt.xlim(0, max(f_mode_2, 1.5) + 0.5) # Zoom in on the relevant frequencies
plt.grid(True)
plt.axvline(f_mode_1, color='green', linestyle='--', label=f'Mode 1 (f1): {f_mode_1:.3f} Hz')
plt.axvline(f_mode_2, color='purple', linestyle='--', label=f'Mode 2 (f2): {f_mode_2:.3f} Hz')
plt.legend()
plt.show() 


# --- Final Summary ---
print("\n--- MEASURED FREQUENCIES (Normal Modes) ---")
print(f"Frequencies when both legs are pulled back (Mode 1): {f_mode_1:.3f} Hz")
print(f"Frequencies when just the bottom leg is excited (Mode 2): {f_mode_2:.3f} Hz")
print(f"The motion is clearly non-sinusoidal as it is composed of two primary frequencies ({f_mode_1:.3f} Hz and {f_mode_2:.3f} Hz), which results in the beating pattern observed in the position-time plot.")
```
